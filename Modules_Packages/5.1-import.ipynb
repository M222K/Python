##### Importing Modules in Python: Modules and Packages
In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16) #math module has various functions to use direclty in our code
```

```python
from math import sqrt,pi #we can import specific function and use directly in our code without using math.
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np #importing the numpy as alias name np 
np.array([1,2,3,4])
```

```python
from math import * #imports all functions in module
print(sqrt(16))
print(pi)
```

Custom Packages- in order to create custom packages,we need to create a package and define our init file after that the files inside packages becomes our module and we can use its methods by importing it.


```python
from package.maths import addition
addition(2,3)
```

```python
from package import maths
maths.addition(2,3)
```

#### Conclusion
Importing modules and packages in Python allows you to organize your code, reuse functionalities, and keep your projects clean and manageable. By understanding how to import modules, specific functions, and use relative imports within packages, you can structure your Python applications more effectively.
