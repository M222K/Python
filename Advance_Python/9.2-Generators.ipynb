#### Generators
Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

Note- Generators are the subclass are iterator itself

```python
def square(n):
    for i in range(3):
        yield i**2 #this yield keyword create a generator object that can be iterated over to get the squared values of i from 0 to 2 and it will store the variables in memory 
```

```python
square(3)
```

```python
#values stored in memory for each i**2 are 0,1,4 now to return these values we can use a for loop to iterate over the generator object created by the square function and print each value one by one

# now this will go to square(3) , then execute loop i==0 , go to yield line then generte i**2=0 save and return it and do this till range(3)

#thus yield keyword is not creating values ek saath par when runned it creates the values and return when called

for i in square(3):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
next(gen) #so basically when you want to use function and iterators together you can use yield keyword to create a generator object that can be iterated over to get the values one by one without storing them all in memory at once. This is especially useful when working with large datasets or infinite sequences.
```

```python
for val in gen:
    print(val)
```

#### Practical Example: Reading Large Files
Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

-helps in reducing the memory load all at once.

```python
### Practical : Reading LArge Files

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line #this will store only one line memory to work with
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```

Iterators and generators are powerful tools in Python for creating and handling sequences of data efficiently. Understanding these concepts will enable you to write more efficient and memory-conscious Python programs.
