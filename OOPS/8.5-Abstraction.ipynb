#### Abstraction
Abstraction is the concept of hiding the complex implementation details and showing only the necessary features of an object. This helps in reducing programming complexity and effort.

example with real world 👍
take example of washing machine ,it has all the feautres and button to start, stop , wash , modes to wash on display , but the actual implementation and coomplexity is not displayed to the user that how that particulat feautre works..

```python
from abc import ABC,abstractmethod

## Abstract base cclass-to make a class as abstract we need to inherit it from ABC and then we can define the abstract method using @abstractmethod decorator and the derived class must implement the abstract method otherwise it will also be considered as abstract class and we can not create object of it.
class Vehicle(ABC):
    def drive(self):
        print("The vehicle is used for driving")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print("Car enginer started")

def operate_vehicle(vehicle):
    vehicle.start_engine()
    vehicle.drive()

car=Car()
operate_vehicle(car)

```
